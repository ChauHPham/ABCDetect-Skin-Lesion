```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, Concatenate
from tensorflow.keras.utils import plot_model

# Referenced: https://github.com/Rajsoni03/neuralplot
# https://keras.io/api/utils/model_plotting_utils/
# Simplified Keras-style UNet based on DL algorithm (for report)

def build_keras_unet(input_shape=(128, 128, 3)):
    inputs = Input(shape=input_shape)
    c1 = Conv2D(64, 3, activation='relu', padding='same')(inputs)
    p1 = MaxPooling2D(2)(c1)

    c2 = Conv2D(128, 3, activation='relu', padding='same')(p1)
    p2 = MaxPooling2D(2)(c2)

    c3 = Conv2D(256, 3, activation='relu', padding='same')(p2)

    u1 = UpSampling2D(2)(c3)
    concat1 = Concatenate()([u1, c2])
    c4 = Conv2D(128, 3, activation='relu', padding='same')(concat1)

    u2 = UpSampling2D(2)(c4)
    concat2 = Concatenate()([u2, c1])
    c5 = Conv2D(64, 3, activation='relu', padding='same')(concat2)

    outputs = Conv2D(1, 1, activation='sigmoid')(c5)
    model = Model(inputs, outputs)
    return model

model = build_keras_unet()
plot_model(model, to_file="keras_unet_architecture.png", show_shapes=True)
model.summary()
```
