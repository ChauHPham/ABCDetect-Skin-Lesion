```python
import os
import shutil
from utils import download_kaggle_dataset, merge_folders, split_dataset, move_all_files
```

Dataset Downloader Hyperparameters

```python
# Kaggle specific - Do not modify
HAM10K_KAGGLE_DATASET_URL = "kmader/skin-cancer-mnist-ham10000"
HAM10K_KAGGLE_SEGMENTATION_MASKS_URL = "tschandl/ham10000-lesion-segmentations"

# User specific - Please modify
DOWNLOAD_PATH = r"C:\Users\colwy\Documents\SFU\Spring2025\CMPT 419\Project\2025_1_project_06\dataset"
HAM10K_DESIRED_DATASET_FOLDER_NAME = "kmader"
HAM10K_DESIRED_SEGMENTATION_MASKS_FOLDER_NAME = "tschandl"
```

The HAM10000 dataset is originally downloaded into two separate folders:
- `part_1`
- `part_2`

This script automates the process by:
1. Downloading the dataset
2. Merging all images from both `part_1` and `part_2` into a single directory
3. Move the metadata such as tables and excel sheets into the desired folder
4. Clean up the folder

📂 **Final destination folders:**
`DOWNLOAD_PATH + HAM10K_DESIRED_DATASET_FOLDER_NAME`
1. `images` folder with all the skin lesions
2. `metadata` folder with the tables

```python
HAM10K_FINAL_IMAGE_PATH = os.path.join(DOWNLOAD_PATH, HAM10K_DESIRED_DATASET_FOLDER_NAME, "images")
HAM10K_FINAL_METADATA_PATH = os.path.join(DOWNLOAD_PATH, HAM10K_DESIRED_DATASET_FOLDER_NAME, "metadata")

if not os.path.exists(HAM10K_FINAL_IMAGE_PATH) or not os.path.exists(HAM10K_FINAL_METADATA_PATH):
    # Download the image files
    HAM10K_dataset_download_location = download_kaggle_dataset(
        HAM10K_KAGGLE_DATASET_URL, DOWNLOAD_PATH, force_download=False)

    # Merge the images from part 1 and 2 into desired destination path
    merge_folders(os.path.join(HAM10K_dataset_download_location,"HAM10000_images_part_1"), HAM10K_FINAL_IMAGE_PATH)
    merge_folders(os.path.join(HAM10K_dataset_download_location,"HAM10000_images_part_2"), HAM10K_FINAL_IMAGE_PATH)

    # Move excel sheets and other metadata to desired folder
    move_all_files(HAM10K_dataset_download_location, HAM10K_FINAL_METADATA_PATH, exclude_subdirectories=True)

    # Remove now empty default download folders
    shutil.rmtree(os.path.join(DOWNLOAD_PATH, "datasets"))
    print("Successfully completed HAM10000 dataset initialization")
else:
    print(f"Datasets already successfully created, delete the {HAM10K_FINAL_IMAGE_PATH} and {HAM10K_FINAL_METADATA_PATH} folders to redownload the datasets")
```

The HAM10000 segmentation dataset is easier to manage, and only contains one folder:
- `HAM10000_segmentations_lesion_tschandl`

This script automates the process by:
1. Downloading the dataset
2. Merging all images from both `HAM10000_segmentations_lesion_tschandl` into a the desired directory
3. Clean up the folder

📂 **Final destination folders:**
`DOWNLOAD_PATH + HAM10K_DESIRED_SEGMENTATION_MASKS_FOLDER_NAME`
1. `masks` folder with all the skin lesion masks

```python
HAM10K_FINAL_MASKS_PATH = os.path.join(DOWNLOAD_PATH, HAM10K_DESIRED_SEGMENTATION_MASKS_FOLDER_NAME, "masks")

if not os.path.exists(HAM10K_FINAL_MASKS_PATH):
    segmentation_mask_path = download_kaggle_dataset(
        HAM10K_KAGGLE_SEGMENTATION_MASKS_URL, DOWNLOAD_PATH, force_download=False)
    
    # Merge the images from part 1 and 2 into desired destination path
    merge_folders(os.path.join(segmentation_mask_path,"HAM10000_segmentations_lesion_tschandl"), HAM10K_FINAL_MASKS_PATH)

    # Remove now empty default download folders
    shutil.rmtree(os.path.join(DOWNLOAD_PATH, "datasets"))
    print("Successfully completed HAM10000 segmentation mask dataset initialization")

else:
    print(f"Datasets already successfully created, delete the {HAM10K_FINAL_MASKS_PATH} folder to redownload the dataset")
```
